Load Dataset

```python
import pandas as pd 
filepath="../dataset/sjc_final.csv"
data=pd.read_csv(filepath,sep=",")
print(data.head(5))
print(data.shape)
print(data.columns)
print(data.dtypes)
```

Kiểm tra Date Range

```python
data["timestamp"]=pd.to_datetime(data["timestamp"])
print(data["timestamp"].dtypes)
min_day=data["timestamp"].min()
max_day=data["timestamp"].max()
print(min_day)
print(max_day)
date_range=max_day-min_day
print(f"Date range: {date_range}")
missing_value=data.isnull().sum()
print(f"Missing data: {missing_value}")
```

Có ngày nào xuất hiện nhiều lần không?
Khoảng cách giữa các ngày liên tiếp có phải luôn là một ngày không?
Có những khoảng thời gian nào không xuất hiện trong dataset?

```python
duplicate_date=data["timestamp"].duplicated().sum()
print(f"Duplicate dates: {duplicate_date}")

data=data.sort_values("timestamp").reset_index(drop=True)
data["date_diff"]=data["timestamp"].diff()
print(data["date_diff"].value_counts().sort_index())

print("Cac khoang lon hon 1 ngay")
gaps=data[data["date_diff"]>pd.Timedelta(days=1)]
print(gaps[["timestamp","date_diff"]])
print(f"Number of gaps: {len(gaps)}")
```

Kiểm tra các khoảng trống có thể ảnh hưởng đến dữ liệu

```python
# Tinh so ngay lich bi thieu trong moi khoang
gaps["missing_days"]=(
    gaps["date_diff"].dt.days-1
)
print(gaps[["timestamp", "date_diff", "missing_days"]])

print("Total missing calendar days:", gaps["missing_days"].sum())
```

Kiểm tra tính hợp lệ của giá vàng

```python
prices=data[["buy_1l","sell_1l"]]
print("Price summary:")
print(prices.describe())

print("Prices <= 0:")
print((prices<0).sum())

print("Buy price > sell price:")
print((data["buy_1l"]>data["sell_1l"]).sum())

print("Buy price > 0:", (data["buy_1l"] > 0).all())
print("Sell price > 0:", (data["sell_1l"] > 0).all())
```
